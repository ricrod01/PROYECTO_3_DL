### Importando librerías

```python
import sys
import time
import pandas as pd
from pathlib import Path
```

### Carga del modelo (Se probó con Adapter)

```python
# Raíz del proyecto.
ROOT = Path.cwd().parent
sys.path.insert(0, str(ROOT))

# Medir el tiempo de carga del modelo.
inicio = time.perf_counter()

from solucion import generar_sql

tiempo_carga = time.perf_counter() - inicio

print(f'Tiempo de carga: {tiempo_carga:.2f} segundos')
```

### Probando consulta

```python
# Cargar datos de validación.
datos = pd.read_excel(ROOT / 'datos' / 'validacion.xlsx')

# Función para leer el esquema.
def obtener_esquema(id_esquema):
    ruta = f'../esquemas/{id_esquema}.sql'

    with open(ruta, 'r', encoding='utf-8') as archivo:
        return archivo.read()

# Agregando el esquema
datos['esquema'] = datos['id_esquema'].apply(obtener_esquema)

# Seleccionando una pregunta
fila = datos.iloc[0]

# Medir el tiempo  de predicción
inicio = time.perf_counter()

sql_predicho = generar_sql(esquema=fila['esquema'], pregunta=fila['prompt'])

tiempo_prediccion = time.perf_counter() - inicio

print('Pregunta:', fila['prompt'])
print('\nSQL generado:\n', sql_predicho)
print('\nSQL de referencia:\n', fila['sql'])
print(f'\nTiempo de inferencia: {tiempo_prediccion:.2f} segundos')
```
